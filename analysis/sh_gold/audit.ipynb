# SH Gold export audit
Run from this notebook folder. Source files are read only. Candidate detection is intentionally conservative; review instrument identity and all media before replay. No performance backtest was run.


````python
from pathlib import Path
import runpy
runpy.run_path(str(Path('analyze_export.py').resolve()), run_name='__main__')
````

````python
import sqlite3
db = sqlite3.connect("audit.sqlite")
db.row_factory = sqlite3.Row
monthly_sql = "WITH months AS (\n SELECT substr(timestamp,1,7) AS month, COUNT(*) AS messages FROM messages GROUP BY 1\n), cohorts AS (\n SELECT substr(timestamp,1,7) AS month, instrument, COUNT(*) AS signals,\n SUM(valid) AS valid_signals, SUM(side='buy') AS buys, SUM(side='sell') AS sells\n FROM parsed_signals GROUP BY 1,2\n), instruments AS (SELECT 'explicit_gold' AS instrument UNION ALL SELECT 'unspecified')\nSELECT months.month, instruments.instrument, COALESCE(cohorts.signals,0) AS signals,\n COALESCE(cohorts.valid_signals,0) AS valid_signals, months.messages,\n COALESCE(cohorts.buys,0) AS buys, COALESCE(cohorts.sells,0) AS sells\nFROM months CROSS JOIN instruments LEFT JOIN cohorts\n ON cohorts.month=months.month AND cohorts.instrument=instruments.instrument\nORDER BY months.month, instruments.instrument;"
[dict(row) for row in db.execute(monthly_sql)]
````
